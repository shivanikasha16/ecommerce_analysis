# E-Commerce Sales & Customer Analytics Using Python

## 1. Introduction

E-commerce generates large amounts of data related to customers, orders, products, sales, marketing, and business operations. Analyzing this data can help identify patterns in customer behavior, sales performance, profitability, and operational efficiency.

This project analyzes an e-commerce dataset using Python to understand sales trends, customer characteristics, profitability, marketing performance, delivery patterns, and returns.

The project focuses on data cleaning, exploratory data analysis, descriptive statistics, statistical analysis, and data visualization.

```python
# Number of unique values in important categorical columns

columns = [
    'return_status',
    'return_reason',
    'campaign_name',
    'delivery_status',
    'marketing_channel'
]

for col in columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(15))
```

```python
# Missing percentage for every column

missing_percentage = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)

missing_percentage
```

```python
df[df['delivery_status'] == 'Cancelled'][
    ['delivery_days',
     'estimated_delivery_days',
     'customer_rating',
     'review_sentiment',
     'customer_review']
].isnull().sum()
```

```python
df.groupby('delivery_status')[
    ['delivery_days', 'estimated_delivery_days', 'customer_rating']
].apply(lambda x: x.isnull().sum())
```

```python
pd.crosstab(
    df['marketing_channel'].fillna('No Channel'),
    df['campaign_name'].fillna('No Campaign')
)
```

```python
campaign_check = df.groupby(
    df['campaign_name'].isna()
)['marketing_channel'].value_counts()

campaign_check
```

```python
df[df['campaign_name'].isna()]['marketing_channel'].value_counts()
```

```python
campaign_by_channel = (
    df.groupby('marketing_channel')['campaign_name']
    .apply(lambda x: x.notna().sum())
)

campaign_by_channel
```

```python
campaign_percentage = (
    df.groupby('marketing_channel')['campaign_name']
    .apply(lambda x: x.notna().mean() * 100)
    .sort_values(ascending=False)
)

campaign_percentage
```

```python
df_clean = df.copy()
```

```python
df_clean['order_date'] = pd.to_datetime(
    df_clean['order_date'],
    errors='coerce'
)
```

```python
df_clean['customer_postal_code'] = (
    df_clean['customer_postal_code'].astype(str)
)
```

```python
df_clean['return_status'] = (
    df_clean['return_status'].fillna('Not Returned')
)

df_clean['return_reason'] = (
    df_clean['return_reason'].fillna('No Return')
)

df_clean['campaign_name'] = (
    df_clean['campaign_name'].fillna('No Specific Campaign')
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean['coupon_code'].value_counts(dropna=False)
```

```python
df_clean.groupby('coupon_code', dropna=False)['discount_amount'].mean().sort_values(ascending=False)
```

```python
df_clean.groupby(
    df_clean['coupon_code'].isna()
)['discount_amount'].agg(['count', 'mean', 'min', 'max'])
```

```python
pd.crosstab(
    df_clean['coupon_code'].isna(),
    df_clean['discount_amount'] == 0
)
```
